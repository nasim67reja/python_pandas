# Sorting Pandas DataFrames

This notebook demonstrates how to sort DataFrames in pandas using different methods, including sorting by index and by values.

## Import Required Libraries

```python
import pandas as pd
```

## Create a Sample DataFrame

```python
data = {
    'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'],
    'Age': [25, 30, 22, 35, 28],
    'Salary': [50000, 55000, 40000, 70000, 48000]
}
df = pd.DataFrame(data)
df
```

## Sort by Index (Row Labels)
Use `sort_index()` to sort the DataFrame by its index.

```python
# Sort by index in ascending order
df_sorted_index = df.sort_index()
df_sorted_index
```

```python
# Sort by index in descending order
df_sorted_index_desc = df.sort_index(ascending=False)
df_sorted_index_desc
```

## Sort by Column Values
Use `sort_values()` to sort the DataFrame by one or more columns.

```python
# Sort by Age in ascending order
df_sorted_age = df.sort_values(by='Age')
df_sorted_age
```

```python
# Sort by Salary in descending order
df_sorted_salary_desc = df.sort_values(by='Salary', ascending=False)
df_sorted_salary_desc
```

```python
# Sort by multiple columns: first by Age, then by Salary
df_sorted_multi = df.sort_values(by=['Age', 'Salary'])
df_sorted_multi
```

## Sort Columns (Axis=1)
You can also sort the columns alphabetically.

```python
df_sorted_columns = df.sort_index(axis=1)
df_sorted_columns
```
