# Pandas DataFrame Index Operations

This notebook demonstrates various index operations in Pandas DataFrames, including setting, resetting, renaming, and using MultiIndex.

## Import Required Libraries

Import the pandas library for DataFrame operations.

```python
import pandas as pd
```

## Create a DataFrame

Create a sample DataFrame to demonstrate index operations.

```python
# Sample data
data = {
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'State': ['NY', 'CA', 'IL', 'TX', 'AZ'],
    'Population': [8419000, 3980000, 2716000, 2328000, 1690000]
}

df = pd.DataFrame(data)
print(df)
```

## Set Index for DataFrame

Use the `set_index()` method to set a specific column as the index.

```python
# Set 'City' as the index
df_indexed = df.set_index('City')
print(df_indexed)
```

## Reset Index

Use the `reset_index()` method to reset the index to the default integer-based index.

```python
# Reset the index back to default
df_reset = df_indexed.reset_index()
print(df)
```

## Change Index Name

Modify the name of the index using the `.name` attribute.

```python
# Set 'State' as index and rename the index
df_state_index = df.set_index('State')
df_state_index.index.name = 'StateName'
df_state_index
```

## Use MultiIndex

Demonstrate the creation and usage of MultiIndex for hierarchical indexing.

```python
# Set both 'State' and 'City' as a MultiIndex
df_multi = df.set_index(['State', 'City'])
df_multi
```

```python
# accessing the row using loc()
print(df_state_index.loc['CA'])  # Accessing data for California
```
