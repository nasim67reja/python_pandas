# Creating Dataframes

## Empty, from list, from numpy array, from dictonary

- pandas.DataFrame(data, index, columns) 

```python
import pandas as pd
```

```python
empty_df=pd.DataFrame()
print(empty_df)
```

```python
# Create a DataFrame from a list
list_data = [1, 2, 3, 4, 5]
df_list = pd.DataFrame(list_data, 
                       index=['a', 'b', 'c', 'd', 'e'],
                       columns=['Values'])
# print(df_list)

# initialize list of lists
data = [['tom', 10], ['nick', 15], ['juli', 14]]

# Create the pandas DataFrame
df = pd.DataFrame(data, columns=['Name', 'Age'],index=['a', 'b', 'c'])

print(df)
```

```python
# Create a DataFrame from a numpy array
import numpy as np
array=np.array([[1,2,3],[4,5,6]])
array=array.reshape(3,2)
# print(array)
df_array=pd.DataFrame(array,columns=['A','B'],index=['a','b','c'])
print(df_array)
```

```python
# Create a DataFrame from a dictionary for algorithmic trading
data = {
    'Timestamp': ['2023-01-01 09:30', '2023-01-01 09:31', '2023-01-01 09:32'],
    'Stock': ['AAPL', 'AAPL', 'AAPL'],
    'Price': [150.25, 150.50, 150.75],
    'Volume': [1000, 1500, 1200]
}
df_trading = pd.DataFrame(data)
print(df_trading)


# Initialize data of lists
data = [{'b': 2, 'c': 3}, {'a': 10, 'b': 20, 'c': 30}]

# Creates pandas DataFrame by passing
# Lists of dictionaries and row index.
df = pd.DataFrame(data, index=['first', 'second'])

print(df)
```

```python
# Create a DataFrame from zip of two lists
Name = ['tom', 'krish', 'nick', 'juli']

Age = [25, 30, 26, 22]

# and merge them by using zip().
list_of_tuples = list(zip(Name, Age))

# Assign data to tuples.
list_of_tuples


# Converting lists of tuples into
# pandas Dataframe.
df = pd.DataFrame(list_of_tuples,
                  columns=['Name', 'Age'])

print(df)
```
