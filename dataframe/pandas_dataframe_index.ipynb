# Indexing and Selecting Data with Pandas

This notebook demonstrates various ways to index and select data in Pandas DataFrames, including column selection, row selection, slicing, boolean indexing, and advanced indexing methods.

## Import Required Libraries

```python
import pandas as pd
```

## Read data from csv

```python
df=pd.read_csv('../data/algo_trading_data.csv',index_col='Date')
df
```

## Selecting Columns
You can select a single column or multiple columns using the indexing operator [].

```python
# df.head(5)
display(df.head(5)[['Open','High','Low','Close']])
```

```python
# Select multiple columns
df[df["Volume"]>1050]
```

## Selecting Rows by Index
Use `iloc` for integer-location based indexing and `loc` for label-based indexing.

```python
# Select the first row
df.iloc[0]
```

```python
# Select rows by label
df.loc[0:2]  # Rows with index 0, 1, 2
```

## Slicing Rows and Columns
You can slice both rows and columns using `loc` and `iloc`.

```python
# Select first three rows and 'Name', 'Age' columns
df.loc[0:2, ['Name', 'Age']]
```

## Boolean Indexing
Filter rows based on a condition.

```python
# Select rows where Age > 25
df[df['Age'] > 25]
```

## Accessing a Single Value
Use `at` for label-based and `iat` for integer position-based access.

```python
# Access the Salary of the row with index 2
df.at[2, 'Salary']
```

```python
# Access the value at row 2, column 3 (Salary)
df.iat[2, 3]
```

# Pandas DataFrame: Merging, Joining, and Concatenating

This notebook demonstrates how to combine DataFrames in pandas using merging, joining, and concatenating methods.

## Import Required Libraries

```python
import pandas as pd
```

## Create Sample DataFrames

```python
data1 = {
    'ID': [1, 2, 3, 4],
    'Name': ['John', 'Alice', 'Bob', 'Eve']
}
data2 = {
    'ID': [3, 4, 5, 6],
    'Salary': [40000, 50000, 60000, 70000]
}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df1, df2
```

## Merging DataFrames
Use `pd.merge()` to combine DataFrames based on a common column.

```python
# Inner join (only matching IDs)
merged_inner = pd.merge(df1, df2, on='ID', how='inner')
merged_inner
```

```python
# Outer join (all IDs, fill missing with NaN)
merged_outer = pd.merge(df1, df2, on='ID', how='outer')
merged_outer
```

## Joining DataFrames
Use `DataFrame.join()` to join on the index.

```python
df3 = df1.set_index('ID')
df4 = df2.set_index('ID')
joined = df3.join(df4, how='inner')
joined
```

## Concatenating DataFrames
Use `pd.concat()` to stack DataFrames vertically or horizontally.

```python
# Concatenate vertically (default axis=0)
concat_vertical = pd.concat([df1, df2], ignore_index=True)
concat_vertical
```

```python
# Concatenate horizontally (axis=1)
concat_horizontal = pd.concat([df1, df2], axis=1)
concat_horizontal
```
