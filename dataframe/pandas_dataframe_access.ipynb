# Pandas DataFrame Access Methods

This notebook demonstrates various ways to access data in a Pandas DataFrame, including columns, rows, multiple rows/columns, filtering, and accessing specific cells using `loc`, `iloc`, `at`, and `iat`.

## Import Required Libraries

```python
import pandas as pd
```

## Create a Sample DataFrame

```python
data = {
    'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'],
    'Age': [25, 30, 22, 35, 28],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'],
    'Salary': [50000, 55000, 40000, 70000, 48000]
}
df = pd.DataFrame(data)
df
```

## Access the Entire DataFrame

Simply refer to the DataFrame variable to display all data.

```python
df
```

## Access a Column

Access a column using bracket notation.

```python
df['Name']
```

## Access a Row by Index (iloc)

Use `iloc` for positional indexing.

```python
print(df.iloc[4])  # Second row
df.iloc[2:4,1:3]  # Second row, third column
```

## Access Multiple Rows or Columns

Select multiple rows or columns by passing lists.

```python
df.loc[0:1,["Name","Salary"]]  # First three rows, Name and Age columns
```

## Access Rows Based on Condition

Filter rows where Age > 25.

```python
df[df['Age'] > 25]
df[df["Gender"] =="Male"]  # Filter rows where Salary is greater than 50000  ]
```

## Access a Specific Cell with at and iat

Use `at` for label-based and `iat` for integer position-based access.

```python
df.at[2, 'Salary']  # Salary of row with index 2
```

```python
df.iat[2, 3]  # Third row, fourth column (Salary)
```
