# Handling Missing Data

### Checking for missing values using isnull()

```python
import pandas as pd
import numpy as np

# Create a demo dataset for algorithmic trading
data = {
    'Date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'Open': [100, 102, np.nan, 105, 107, np.nan, 110, 112, 113, np.nan],
    'High': [101, 103, 104, 106, 108, 109, 111, 113, 114, 115],
    'Low': [99, 101, 102, 104, np.nan, 108, 109, 111, np.nan, 114],
    'Close': [100.5, 102.5, 103.5, np.nan, 107.5, 108.5, 110.5, 112.5, 113.5, 114.5],
    'Volume': [1000, 1100, 1200, 1300, 1400, np.nan, 1600, 1700, 1800, 1900]
}

# Create a DataFrame
df=pd.DataFrame(data)
mv=df.isnull()
mv=df.notnull()
df
```

### 2.filtering data based on missing values

```python
# Filter rows where 'Open' column has missing values
missing_open = df[df['Volume'].notnull()]
print(missing_open)
print(df)
```

## Filling Missing Values in Pandas Using fillna(), replace() and interpolate()

```python
# using filna()

fill_with_zero=df.fillna(0)
fill_with_zero
# df.fillna(method='bfill')
```

### Using replace()

```python
df.replace(to_replace=np.nan, value=-99)
# df
```

## Dropping Missing Values in Pandas Using dropna()

```python
dict = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, np.nan, 45, 56],
        'Third Score': [52, 40, 80, 98],
        'Fourth Score': [np.nan, np.nan, np.nan, np.nan]}
df = pd.DataFrame(dict)

# Drop rows with at least one missing value
df.dropna() # Drop rows with at least one missing value
df.dropna(how='all') # Drop rows with all missing values
df.dropna(thresh=2) # Drop rows with at least 2 non-NA values
df.dropna(axis=1)
```
